```python
DATA_PATH = Path("C:\\Users\\This Pc\\OneDrive\\Desktop\\crdb-customer-segmentation\\crdb-customer-segmentation\\data\\customers_synthetic.csv")

df = pd.read_csv(DATA_PATH)
print(df.shape)
df.head()
df.info()
df.describe()
```

```python
print("Missing values per column:")
print(df.isna().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
FEATURES = [
    "balance",
    "monthly_transactions",
    "tenure_years",
    "digital_usage_score",
]

FEATURES = [f for f in FEATURES if f in df.columns]
print("Using features:", FEATURES)

X = df[FEATURES].copy()
X.head()
```

```python
fig, axes = plt.subplots(1, len(FEATURES), figsize=(5 * len(FEATURES), 4))
if len(FEATURES) == 1:
    axes = [axes]

for ax, col in zip(axes, FEATURES):
    sns.histplot(X[col], kde=True, ax=ax, color="#4C9A94")
    ax.set_title(col)

plt.tight_layout()
plt.show()
```
